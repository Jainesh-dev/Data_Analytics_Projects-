# 📂 Dataset Description

Dataset Name:
NASA CMAPSS Turbofan Engine Degradation Dataset (FD001)

Source:
NASA Prognostics Center of Excellence

Description:
This dataset contains operational settings and sensor measurements collected from simulated aircraft engines. It is widely used for predictive maintenance and Remaining Useful Life (RUL) analysis.

Business Objective:
Analyze engine operational data to identify degradation patterns and support maintenance decision-making.

## Key Features

- Engine_ID: Unique identifier for each aircraft engine.
- Cycle: Number of operational cycles completed.
- Operational Settings: Three variables representing operating conditions.
- Sensor Measurements: Twenty-one sensor readings monitoring engine health.
