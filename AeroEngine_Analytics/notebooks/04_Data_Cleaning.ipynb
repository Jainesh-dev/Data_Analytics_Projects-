# Data Cleaning

## Objective

The objective of this notebook is to prepare the dataset for exploratory analysis and predictive maintenance.

Data cleaning decisions are based on the findings from the Data Profiling notebook.

Cleaning activities include:

- Removing constant features
- Verifying data quality
- Preserving important operational information
- Saving the cleaned dataset for downstream analysis

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    "../data/raw/train_FD001.txt",
    sep=r"\s+",
    header=None
)

column_names = [
    "Engine_ID",
    "Cycle",
    "Operational_Setting_1",
    "Operational_Setting_2",
    "Operational_Setting_3",
]

for i in range(1,22):
    column_names.append(f"Sensor_{i}")

df.columns = column_names
```

```python
df.shape
```

### Observation

The original dataset contains all operational settings and sensor measurements before preprocessing.

```python
constant_features = [
    "Operational_Setting_3",
    "Sensor_1",
    "Sensor_5",
    "Sensor_10",
    "Sensor_16",
    "Sensor_18",
    "Sensor_19"
]

df.drop(columns=constant_features, inplace=True)
```

```python
df.shape
```

### Observation

Seven constant features were removed because they contain no variation and therefore provide no useful analytical or predictive information.

```python
df.isnull().sum().sum()
```

### Observation

The cleaned dataset contains no missing values.

```python
df.duplicated().sum()
```

### Observation

No duplicate records remain after cleaning.

```python
df.dtypes
```

### Observation

All variables retain appropriate numeric data types required for statistical analysis and machine learning.

```python
df.to_csv(
    "../data/processed/aeroengine_cleaned.csv",
    index=False
)
df.head()
```
