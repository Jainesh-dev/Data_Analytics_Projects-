# Phase 3: Data Profiling

## Objective

Evaluate the overall quality and structure of the dataset before performing any data cleaning or exploratory data analysis.

## Business Context

Before performing any data cleaning or analysis, it is essential to evaluate the quality of the dataset. Data profiling helps identify missing values, duplicate records, constant features, and potential data quality issues that could impact business decisions and analytical accuracy.

```python
import pandas as pd 
df=pd.read_csv("../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv")
df.isnull().sum()
```

### Observation

No missing values were identified across any feature. This indicates that the dataset is complete and does not require missing value treatment before analysis.

```python
df.duplicated().sum()
```

### Observation

No duplicate employee records were identified, indicating that each employee is represented only once in the dataset.

```python
for col in df.columns:
    if df[col].nunique() == 1:
        print(col)
```

### Observation

Three columns contain only one unique value for every employee. These features do not provide analytical value and will be removed during the data cleaning phase.

```python
df.describe()
```

### Observation

The numerical summary indicates that employee ages, monthly income, and years of experience fall within expected ranges. No obvious anomalies are observed at this stage.

```python
print(df["Attrition"].value_counts())
print(df["Gender"].value_counts())
print(df["Department"].value_counts())
```

Most employees remain with the organization, while a smaller proportion have left. This confirms that employee retention is higher than attrition.
The dataset contains both male and female employees, enabling gender-based workforce analysis.
Employees are distributed across multiple departments, allowing department-wise attrition analysis in later stages.
