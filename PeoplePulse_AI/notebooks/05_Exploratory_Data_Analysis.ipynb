# Phase 5: Exploratory Data Analysis (EDA)

## Objective

Explore the cleaned employee dataset to identify patterns, relationships, and factors associated with employee attrition. The findings from this analysis will support HR leaders in making data-driven workforce decisions.

## Business Context

Following data cleaning, the dataset is now ready for exploratory analysis. The objective of this phase is to uncover workforce trends, identify potential drivers of employee attrition, and generate actionable business insights for HR leadership.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/employee_attrition_cleaned.csv")
sns.set_style("whitegrid")
df.head(10)
```

```python
df["Attrition"].value_counts(normalize=True)*100
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Attrition", data=df)
plt.title("Employee Attrition Distribution")
plt.show()
```

```python
pd.crosstab(df["Department"], df["Attrition"])
sns.countplot(data=df,x="Department",hue="Attrition")
plt.title("Department Attrition Data")
plt.show()
```

```python
pd.crosstab(df["JobRole"],df["Attrition"])
sns.countplot(data=df,x="JobRole",hue="Attrition")
plt.title("Job Role Attrition Analysis")
plt.show()
```

```python

sns.boxplot(data=df,y="MonthlyIncome",x="Attrition")
plt.title("Income Attrition")
plt.show()
```

```python

sns.countplot(data=df,x="JobSatisfaction",hue="Attrition")
plt.title("Satisfactory Attrition")
plt.show()
```

```python
pd.crosstab(df["WorkLifeBalance"],df["Attrition"])
sns.countplot(data=df,x="WorkLifeBalance",hue="Attrition")
plt.title("Work Life Balance Attrition")
plt.show()
```

```python
pd.crosstab(df["YearsAtCompany"],df["Attrition"])
sns.boxplot(data=df,y="YearsAtCompany",x="Attrition")
plt.title("YearsAtCompany Attrition")
plt.show()
```

```python
sns.countplot(data=df, x="OverTime", hue="Attrition")
```

```python
sns.countplot(data=df, x="Gender", hue="Attrition")
```

```python
corr = df.select_dtypes(include="number").corr()
plt.figure(figsize=(12,8))
sns.heatmap(corr, cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

📋 Key Findings


Approximately 16% of employees left the organization.
Overtime appears strongly associated with employee attrition.
Lower monthly income is linked with higher employee turnover.
Employees with lower job satisfaction exhibit higher attrition.
Longer employee tenure is generally associated with better retention.

📌 Business Recommendations

Review overtime policies.
Improve employee engagement.
Monitor lower-income employee retention.
Conduct retention interviews.
Strengthen work-life balance initiatives.
